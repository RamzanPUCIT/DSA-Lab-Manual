## 📖 Part A: Basic Terms

| Term | Meaning | Example |
|---|---|---|
| Ascending / descending | Smallest first / largest first | [1, 3, 7] / [7, 3, 1] |
| Comparison | Checking which of two elements is bigger | `a[j] > a[j+1]` |
| Swap | Exchanging two elements | `a[i], a[j] = a[j], a[i]` |
| Pass | One complete scan over the (unsorted part of the) array | - |
| In-place | Uses only O(1) extra memory | Bubble, Selection, Insertion |
| Stable | Equal elements keep their original relative order | Two students with 80 marks stay in roll-number order |

## 🫧 Part B: Bubble Sort

**Idea:** repeatedly compare **adjacent** elements and swap them if they are in the wrong order. After each pass, the **largest remaining element “bubbles up”** to its correct position at the end.

**Pseudocode**

```text
ALGORITHM BubbleSort(A)
  FOR i ← 0 TO n − 2 DO                  // pass number
      swapped ← FALSE
      FOR j ← 0 TO n − 2 − i DO           // last i elements already sorted
          IF A[j] > A[j + 1] THEN
              swap A[j], A[j + 1] ; swapped ← TRUE
      IF swapped = FALSE THEN BREAK       // already sorted: stop early
```

#### Dry run on [5, 1, 4, 2, 8]

| Pass | Array after the pass | Fixed at the end |
|---|---|---|
| 1 | [1, 4, 2, 5, 8] | 8 |
| 2 | [1, 2, 4, 5, 8] | 5, 8 |
| 3 | [1, 2, 4, 5, 8] (no swaps) | Stop early ✅ |

````python
def bubble_sort(arr):
    a = arr[:]                           # work on a copy
    n = len(a)
    for i in range(n - 1):
        swapped = False
        for j in range(n - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True
        print(f"Pass {i + 1}: {a}")
        if not swapped:
            break
    return a

bubble_sort([5, 1, 4, 2, 8])
````

## 👆 Part C: Selection Sort

**Idea:** find the **minimum** element in the unsorted part and **put it at the front** (swap it with the first unsorted element). Repeat for the rest. Like choosing the shortest student and sending them to the front of the line, again and again.

**Pseudocode**

```text
ALGORITHM SelectionSort(A)
  FOR i ← 0 TO n − 2 DO
      min_index ← i
      FOR j ← i + 1 TO n − 1 DO
          IF A[j] < A[min_index] THEN min_index ← j
      swap A[i], A[min_index]
```

#### Dry run on [64, 25, 12, 22, 11]

| i | Minimum found | Array after swap |
|---|---|---|
| 0 | 11 | [11, 25, 12, 22, 64] |
| 1 | 12 | [11, 12, 25, 22, 64] |
| 2 | 22 | [11, 12, 22, 25, 64] |
| 3 | 25 | [11, 12, 22, 25, 64] |

````python
def selection_sort(arr):
    a = arr[:]
    n = len(a)
    for i in range(n - 1):
        min_index = i
        for j in range(i + 1, n):
            if a[j] < a[min_index]:
                min_index = j
        a[i], a[min_index] = a[min_index], a[i]
        print(f"i = {i}: {a}")
    return a

selection_sort([64, 25, 12, 22, 11])
````

> **💡 Selection sort’s special property**
>
> It makes **at most n − 1 swaps**, the fewest of all three. This is useful when writing to memory is expensive (e.g., flash memory or EEPROM in embedded devices).

## 🃏 Part D: Insertion Sort

**Idea:** build the sorted part one element at a time. Take the next element (the **key**), **shift** bigger elements of the sorted part one place to the right, and **insert** the key into the gap. This is exactly how you arrange playing cards in your hand.

**Pseudocode**

```text
ALGORITHM InsertionSort(A)
  FOR i ← 1 TO n − 1 DO
      key ← A[i] ; j ← i − 1
      WHILE j ≥ 0 AND A[j] > key DO
          A[j + 1] ← A[j]          // shift right
          j ← j − 1
      A[j + 1] ← key               // insert
```

#### Dry run on [12, 11, 13, 5, 6]

| i | key | Array after inserting key | Sorted part |
|---|---|---|---|
| 1 | 11 | [11, 12, 13, 5, 6] | [11, 12] |
| 2 | 13 | [11, 12, 13, 5, 6] | [11, 12, 13] |
| 3 | 5 | [5, 11, 12, 13, 6] | [5, 11, 12, 13] |
| 4 | 6 | [5, 6, 11, 12, 13] | all |

````python
def insertion_sort(arr):
    a = arr[:]
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]              # shift
            j -= 1
        a[j + 1] = key                   # insert
        print(f"i = {i}, key = {key:>2}: {a}")
    return a

insertion_sort([12, 11, 13, 5, 6])
````

## 📊 Part E: Analysis and Comparison

### Counting the Work

For bubble and selection sort, the inner loop runs (n−1) + (n−2) + … + 1 = **n(n−1)/2** times in the worst case, which is **O(n²)**. For insertion sort, the worst case (reverse-sorted input) also shifts n(n−1)/2 times, but on **already sorted** input the while loop never runs, so it is **O(n)**.

| Algorithm | Best | Average | Worst | Space | Stable? | Swaps (worst) |
|---|---|---|---|---|---|---|
| Bubble (early stop) | O(n) | O(n²) | O(n²) | O(1) | Yes | O(n²) |
| Selection | O(n²) | O(n²) | O(n²) | O(1) | No | O(n) |
| Insertion | O(n) | O(n²) | O(n²) | O(1) | Yes | O(n²) shifts |

````python
import random

def count_ops(sort_name, arr):
    a = arr[:]; n = len(a); comps = writes = 0
    if sort_name == "bubble":
        for i in range(n - 1):
            swapped = False
            for j in range(n - 1 - i):
                comps += 1
                if a[j] > a[j + 1]:
                    a[j], a[j + 1] = a[j + 1], a[j]; writes += 1; swapped = True
            if not swapped:
                break
    elif sort_name == "selection":
        for i in range(n - 1):
            m = i
            for j in range(i + 1, n):
                comps += 1
                if a[j] < a[m]:
                    m = j
            if m != i:
                a[i], a[m] = a[m], a[i]; writes += 1
    else:
        for i in range(1, n):
            key, j = a[i], i - 1
            while j >= 0:
                comps += 1
                if a[j] <= key:
                    break
                a[j + 1] = a[j]; writes += 1; j -= 1
            a[j + 1] = key
    assert a == sorted(arr)
    return comps, writes

n = 500
inputs = {"random": random.sample(range(n), n), "sorted": list(range(n)), "reversed": list(range(n, 0, -1))}
print(f"{'input':<9} {'algorithm':<10} {'comparisons':>12} {'swaps/shifts':>13}")
for name, data in inputs.items():
    for alg in ["bubble", "selection", "insertion"]:
        c, w = count_ops(alg, data)
        print(f"{name:<9} {alg:<10} {c:>12,} {w:>13,}")
````

Look at the “sorted” rows: bubble and insertion do only n − 1 comparisons, while selection still does n(n−1)/2. That is why insertion sort is the champion for nearly-sorted data.

### Stability: Why It Matters

Suppose the student list is already in roll-number order, and we sort by marks. A **stable** sort keeps students with equal marks in roll-number order; an unstable sort may mix them up.

````python
students = [("Ali", 80), ("Sara", 95), ("Hamza", 80), ("Zain", 70)]

def selection_sort_by_marks(recs):
    a = recs[:]
    for i in range(len(a) - 1):
        m = i
        for j in range(i + 1, len(a)):
            if a[j][1] < a[m][1]:
                m = j
        a[i], a[m] = a[m], a[i]
    return a

print("Selection (unstable):", selection_sort_by_marks(students))
print("sorted()  (stable)  :", sorted(students, key=lambda s: s[1]))
````

## 🔬 Bonus: Watch the Algorithms Race

````python
# Bonus: run after completing the tasks
import random, time

def bubble(a):
    a = a[:]; n = len(a)
    for i in range(n - 1):
        sw = False
        for j in range(n - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]; sw = True
        if not sw: break
    return a

def selection(a):
    a = a[:]
    for i in range(len(a) - 1):
        m = min(range(i, len(a)), key=a.__getitem__)
        a[i], a[m] = a[m], a[i]
    return a

def insertion(a):
    a = a[:]
    for i in range(1, len(a)):
        k, j = a[i], i - 1
        while j >= 0 and a[j] > k:
            a[j + 1] = a[j]; j -= 1
        a[j + 1] = k
    return a

data = random.sample(range(10_000), 2_000)
for f in (bubble, selection, insertion, sorted):
    t = time.perf_counter(); f(data); print(f"{f.__name__:<10} {time.perf_counter() - t:.4f} s")
````
