## 🔎 Part A: Linear Search

**Linear (sequential) search** checks every element one by one until it finds the target or reaches the end. It works on **any** array, sorted or not.

**Pseudocode**

```text
ALGORITHM LinearSearch(A, target)
  FOR i ← 0 TO n − 1 DO
      IF A[i] = target THEN RETURN i
  RETURN −1
```

````python
def linear_search(arr, target):
    for i, value in enumerate(arr):
        if value == target:
            return i
    return -1

roll_numbers = [1043, 1017, 1088, 1002, 1056]
print(linear_search(roll_numbers, 1088), linear_search(roll_numbers, 9999))
````

| Case | Steps | Big-O |
|---|---|---|
| Best (first element) | 1 | O(1) |
| Average | ≈ n/2 | O(n) |
| Worst (last / absent) | n | O(n) |
| Extra space | - | O(1) |

## 🎯 Part B: Binary Search

If the array is **sorted**, we can do much better. Look at the **middle** element: if it is the target, done; if the target is smaller, it must be in the **left half**; otherwise in the **right half**. Each step **throws away half** of the remaining elements.

> **⚠️ Precondition**
>
> Binary search works **only on sorted data**. On unsorted data it gives wrong answers silently.

**Pseudocode**

```text
ALGORITHM BinarySearch(A, target)
  low ← 0 ; high ← n − 1
  WHILE low ≤ high DO
      mid ← (low + high) // 2
      IF A[mid] = target THEN RETURN mid
      ELSE IF A[mid] < target THEN low ← mid + 1     // go right
      ELSE high ← mid − 1                            // go left
  RETURN −1
```

#### Dry run: search 23 in [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]

| Step | low | high | mid | A[mid] | Decision |
|---|---|---|---|---|---|
| 1 | 0 | 9 | 4 | 16 | 16 < 23 → go right, low = 5 |
| 2 | 5 | 9 | 7 | 56 | 56 > 23 → go left, high = 6 |
| 3 | 5 | 6 | 5 | 23 | Found at index 5 ✅ |

````python
def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

data = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
print(binary_search(data, 23), binary_search(data, 7))
````

### Why O(log n)?

After 1 step n/2 elements remain, after 2 steps n/4, after k steps n/2ᵏ. We stop when one element remains: n/2ᵏ = 1, so **k = log₂ n**. For 1 crore (10⁷) sorted roll numbers, binary search needs at most **24** comparisons; linear search may need 10,000,000.

````python
import math
for n in [100, 10_000, 10_000_000, 8_000_000_000]:
    print(f"n = {n:>14,}: linear worst = {n:>14,}   binary worst = {math.floor(math.log2(n)) + 1}")
````

### Binary Search Variants

Many problems need not just “is it present?” but **where** exactly. The trick is: when you find a match, **do not stop**; record it and keep searching in one direction.

````python
def first_occurrence(arr, target):
    low, high, answer = 0, len(arr) - 1, -1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            answer = mid          # remember it...
            high = mid - 1        # ...but keep looking on the LEFT
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return answer

def lower_bound(arr, target):
    """First index where arr[i] >= target (= insert position)."""
    low, high = 0, len(arr)       # note: high = len(arr)
    while low < high:
        mid = (low + high) // 2
        if arr[mid] < target:
            low = mid + 1
        else:
            high = mid
    return low

grades = [50, 60, 60, 60, 75, 80, 95]
print(first_occurrence(grades, 60), lower_bound(grades, 70), lower_bound(grades, 100))
````

### Binary Search on the Answer

Sometimes we binary search over **possible answers** instead of an array. Example: the integer square root of n is the largest x with x² ≤ n. Answers lie between 0 and n, and the condition “x² ≤ n” is True for small x and False for large x, so we can binary search it.

````python
def int_sqrt(n):
    low, high, ans = 0, n, 0
    while low <= high:
        mid = (low + high) // 2
        if mid * mid <= n:
            ans = mid
            low = mid + 1
        else:
            high = mid - 1
    return ans

print(int_sqrt(10), int_sqrt(144), int_sqrt(2_000_000_000))
````

## 👉👈 Part C: Two Pointers

The **two-pointer** technique uses two indexes that move through the array, usually one from the **start** and one from the **end**, or both from the start at different speeds. It often turns an O(n²) “check all pairs” solution into **O(n)**.

#### Example: pair with a given sum in a sorted array

You have Rs. 1,000 and a **sorted** price list. Find two items that cost exactly Rs. 1,000. If the sum is too small, move the left pointer right (bigger price). If too large, move the right pointer left.

````python
def pair_with_sum(prices, budget):
    left, right = 0, len(prices) - 1
    while left < right:
        s = prices[left] + prices[right]
        if s == budget:
            return prices[left], prices[right]
        elif s < budget:
            left += 1               # need a bigger sum
        else:
            right -= 1              # need a smaller sum
    return None

print(pair_with_sum([120, 250, 300, 450, 700, 880], 1000))
````

## 🪟 Part D: Sliding Window

A **sliding window** is a range `[left, right]` that moves over the array. Instead of recomputing everything for every window, we **add the element entering** and **remove the element leaving**. This makes many O(n·k) problems **O(n)**.

### Fixed-Size Window

**Problem:** daily sales of a shop for 10 days are given. Find the best total over any **3 consecutive days**.

**Idea**

```text
Sales:   [4, 2, 12, 3, 8, 5, 1, 9, 6, 2]     k = 3
Window 1: [4, 2, 12] = 18
Slide  : remove 4, add 3  -> [2, 12, 3] = 17
Slide  : remove 2, add 8  -> [12, 3, 8] = 23   <- best
...
```

````python
def best_k_days(sales, k):
    window = sum(sales[:k])
    best = window
    for right in range(k, len(sales)):
        window += sales[right] - sales[right - k]   # add new, remove old
        best = max(best, window)
    return best

print(best_k_days([4, 2, 12, 3, 8, 5, 1, 9, 6, 2], 3))
````

### Variable-Size Window

**Problem:** length of the longest substring without repeating characters. Expand `right` one character at a time; whenever a repeat appears, shrink from `left` until the window is valid again.

````python
def longest_unique(s):
    seen = set()
    left = best = 0
    for right, ch in enumerate(s):
        while ch in seen:            # shrink until ch is no longer repeated
            seen.remove(s[left])
            left += 1
        seen.add(ch)
        best = max(best, right - left + 1)
    return best

print(longest_unique("abcabcbb"), longest_unique("pakistan"))
````

Each character enters the window once and leaves at most once, so the total work is **O(n)** even though there is a loop inside a loop.

#include <algorithm>
using namespace std;

int binarySearch(const vector<int>& a, int target) {
    int low = 0, high = (int)a.size() - 1;
    while (low <= high) {
        int mid = low + (high - low) / 2;   // avoids integer overflow
        if (a[mid] == target) return mid;
        if (a[mid] < target) low = mid + 1;
        else high = mid - 1;
    }
    return -1;

## 🔬 Bonus: Linear vs Binary Search Race

````python
# Bonus: run after completing the tasks
import random, time, bisect

n = 1_000_000
data = list(range(0, 2 * n, 2))            # sorted even numbers
queries = [random.randrange(0, 2 * n) for _ in range(200)]

t = time.perf_counter()
for q in queries:
    _ = q in data                          # linear search inside Python
linear_time = time.perf_counter() - t

t = time.perf_counter()
for q in queries:
    i = bisect.bisect_left(data, q)
    _ = i < len(data) and data[i] == q     # binary search
binary_time = time.perf_counter() - t

print(f"200 searches in 1 million items -> linear: {linear_time:.3f}s, binary: {binary_time:.5f}s")
print(f"Binary search was about {linear_time / binary_time:,.0f}x faster")
````
