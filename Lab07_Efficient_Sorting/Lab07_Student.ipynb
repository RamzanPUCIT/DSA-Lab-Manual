## 🔀 Part A: Merge Sort

**Idea:** an array of size 1 is already sorted. So: **divide** the array into two halves, **sort each half recursively**, then **merge** the two sorted halves into one sorted array.

**Merge sort on [38, 27, 43, 3, 9, 82, 10]**

```text
             [38, 27, 43, 3, 9, 82, 10]
             /                        \
    [38, 27, 43, 3]              [9, 82, 10]          DIVIDE
     /          \                  /       \
 [38, 27]    [43, 3]          [9, 82]     [10]
  /   \       /   \           /   \
[38] [27]  [43]  [3]        [9]  [82]                 single elements
  \   /       \   /           \   /
 [27, 38]    [3, 43]          [9, 82]     [10]        MERGE
     \          /                  \       /
    [3, 27, 38, 43]              [9, 10, 82]
             \                        /
             [3, 9, 10, 27, 38, 43, 82]
```

### The Merge Step

You already wrote `merge_sorted` in Lab 03: compare the fronts of two sorted lists and repeatedly take the smaller one. It takes **O(n)** time for a total of n elements.

````python
def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:            # <= keeps merge sort STABLE
            result.append(left[i]); i += 1
        else:
            result.append(right[j]); j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

def merge_sort(arr):
    if len(arr) <= 1:                      # base case
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])           # sort left half
    right = merge_sort(arr[mid:])          # sort right half
    return merge(left, right)              # combine

print(merge_sort([38, 27, 43, 3, 9, 82, 10]))
````

### Why O(n log n)?

Look at the recursion tree. At **every level**, all n elements are merged once, which costs **O(n)** per level. The array is halved at each level, so there are **log₂ n levels**. Total: **n × log n**. This holds in the best, average and worst case.

| Level | Sub-arrays | Size of each | Merge work at this level |
|---|---|---|---|
| 0 | 1 | n | n |
| 1 | 2 | n/2 | n |
| 2 | 4 | n/4 | n |
| … | … | … | … |
| log n | n | 1 | n |

| Property | Merge Sort |
|---|---|
| Time (all cases) | O(n log n) |
| Extra space | O(n) for merging |
| Stable? | Yes |
| In-place? | No |
| Great for | Linked lists, external (disk) sorting, when stability is needed |

## ⚡ Part B: Quick Sort

**Idea:** choose a **pivot** element. **Partition** the array so that everything smaller than the pivot is on its left and everything larger is on its right. Now the pivot is in its **final position**. Recursively sort the left and right parts. No merge step is needed.

### Lomuto Partition

**Pseudocode**

```text
ALGORITHM Partition(A, low, high)        // pivot = A[high]
  pivot ← A[high] ; i ← low − 1           // i = end of the "smaller" region
  FOR j ← low TO high − 1 DO
      IF A[j] ≤ pivot THEN
          i ← i + 1 ; swap A[i], A[j]
  swap A[i + 1], A[high]                  // put pivot in its final place
  RETURN i + 1
```

#### Dry run: partition [10, 80, 30, 90, 40, 50, 70] with pivot 70

| j | A[j] | A[j] ≤ 70? | i | Array |
|---|---|---|---|---|
| 0 | 10 | Yes → swap A[0], A[0] | 0 | [10, 80, 30, 90, 40, 50, 70] |
| 1 | 80 | No | 0 | [10, 80, 30, 90, 40, 50, 70] |
| 2 | 30 | Yes → swap A[1], A[2] | 1 | [10, 30, 80, 90, 40, 50, 70] |
| 3 | 90 | No | 1 | [10, 30, 80, 90, 40, 50, 70] |
| 4 | 40 | Yes → swap A[2], A[4] | 2 | [10, 30, 40, 90, 80, 50, 70] |
| 5 | 50 | Yes → swap A[3], A[5] | 3 | [10, 30, 40, 50, 80, 90, 70] |
| end | - | swap A[4], pivot | - | [10, 30, 40, 50, **70**, 90, 80] |

````python
def partition(a, low, high):
    pivot = a[high]
    i = low - 1
    for j in range(low, high):
        if a[j] <= pivot:
            i += 1
            a[i], a[j] = a[j], a[i]
    a[i + 1], a[high] = a[high], a[i + 1]
    return i + 1

def quick_sort(a, low=0, high=None):
    if high is None:
        high = len(a) - 1
    if low < high:
        p = partition(a, low, high)        # pivot is now at index p
        quick_sort(a, low, p - 1)          # sort left of pivot
        quick_sort(a, p + 1, high)         # sort right of pivot
    return a

print(quick_sort([10, 80, 30, 90, 40, 50, 70]))
````

### Best, Average and Worst Case

| Case | When | Recursion depth | Time |
|---|---|---|---|
| Best / average | Pivot splits the array roughly in half | log n | O(n log n) |
| Worst | Pivot is always the smallest or largest (e.g., already sorted input with last-element pivot) | n | O(n²) |

**Fix:** choose a **random pivot** (or the median of three). Then the bad case becomes extremely unlikely for any input.

````python
import random, sys

def quick_sort_random(a, low=0, high=None):
    if high is None:
        high = len(a) - 1
    if low < high:
        r = random.randint(low, high)      # random pivot
        a[r], a[high] = a[high], a[r]
        p = partition(a, low, high)
        quick_sort_random(a, low, p - 1)
        quick_sort_random(a, p + 1, high)
    return a

sys.setrecursionlimit(10_000)
sorted_input = list(range(2000))
print(quick_sort_random(sorted_input[:])[:10], "... works fast even on sorted input")
````

| Property | Quick Sort |
|---|---|
| Time | O(n log n) average, O(n²) worst |
| Extra space | O(log n) recursion stack (average) |
| Stable? | No |
| In-place? | Yes |
| Great for | General-purpose in-memory sorting; very fast in practice (cache-friendly) |

## 🔢 Part C: Counting Sort (Non-Comparison)

Any algorithm that sorts only by **comparing** elements needs at least about **n log n** comparisons in the worst case. But if the values are **small integers in a known range** (e.g., marks 0–100, ages 0–120), we can avoid comparisons: just **count** how many times each value occurs.

````python
def counting_sort(arr, max_value):
    count = [0] * (max_value + 1)
    for x in arr:                          # count occurrences: O(n)
        count[x] += 1
    result = []
    for value in range(max_value + 1):     # rebuild: O(k)
        result.extend([value] * count[value])
    return result

quiz_marks = [7, 3, 10, 3, 8, 7, 7, 0, 10, 5]
print(counting_sort(quiz_marks, 10))
````

Time **O(n + k)** where k is the range of values. Space O(k). It beats O(n log n) when k is not much larger than n. Counting sort is the building block of **Radix Sort**, which sorts large numbers digit by digit.

## 🎯 Part D: Quickselect: The k-th Smallest Without Full Sorting

To find the median or the 3rd highest mark, sorting costs O(n log n). **Quickselect** uses partition but recurses into **only one side**, giving **O(n) on average**.

````python
def quickselect(a, k):                     # k-th smallest, k starts at 1
    a = a[:]
    low, high = 0, len(a) - 1
    while True:
        r = random.randint(low, high)
        a[r], a[high] = a[high], a[r]
        p = partition(a, low, high)
        if p == k - 1:
            return a[p]
        elif p < k - 1:
            low = p + 1                    # answer is on the right
        else:
            high = p - 1                   # answer is on the left

marks = [67, 89, 45, 92, 78, 55, 81]
print("median:", quickselect(marks, 4), "| 2nd smallest:", quickselect(marks, 2))
````

## 📋 Part E: The Complete Sorting Comparison

| Algorithm | Best | Average | Worst | Space | Stable | Notes |
|---|---|---|---|---|---|---|
| Bubble | O(n) | O(n²) | O(n²) | O(1) | Yes | Teaching only |
| Selection | O(n²) | O(n²) | O(n²) | O(1) | No | Fewest swaps |
| Insertion | O(n) | O(n²) | O(n²) | O(1) | Yes | Small / nearly sorted |
| Merge | O(n log n) | O(n log n) | O(n log n) | O(n) | Yes | Guaranteed, linked lists |
| Quick | O(n log n) | O(n log n) | O(n²) | O(log n) | No | Fastest in practice |
| Heap (Lab 15) | O(n log n) | O(n log n) | O(n log n) | O(1) | No | Guaranteed + in-place |
| Counting | O(n + k) | O(n + k) | O(n + k) | O(k) | Yes* | Small integer range |
| Timsort (Python) | O(n) | O(n log n) | O(n log n) | O(n) | Yes | Real-world default |

## ⚙️ C++ Corner

**C++**

```cpp
#include <iostream>
#include <vector>
#include <algorithm>
using namespace std;

void mergeSort(vector<int>& a, int l, int r) {
    if (l >= r) return;
    int m = l + (r - l) / 2;
    mergeSort(a, l, m);
    mergeSort(a, m + 1, r);
    vector<int> tmp;
    int i = l, j = m + 1;
    while (i <= m && j <= r) tmp.push_back(a[i] <= a[j] ? a[i++] : a[j++]);
    while (i <= m) tmp.push_back(a[i++]);
    while (j <= r) tmp.push_back(a[j++]);
    for (int k = 0; k < (int)tmp.size(); k++) a[l + k] = tmp[k];
}

int partition(vector<int>& a, int low, int high) {
    int pivot = a[high], i = low - 1;
    for (int j = low; j < high; j++)
        if (a[j] <= pivot) swap(a[++i], a[j]);
    swap(a[i + 1], a[high]);
    return i + 1;
}

void quickSort(vector<int>& a, int low, int high) {
    if (low < high) {
        int p = partition(a, low, high);
        quickSort(a, low, p - 1);
        quickSort(a, p + 1, high);
    }
}

int main() {

    mergeSort(a, 0, a.size() - 1);
    quickSort(b, 0, b.size() - 1);
    sort(b.begin(), b.end());          // the library version (Introsort)
    for (int x : a) cout << x << " ";
}
```
