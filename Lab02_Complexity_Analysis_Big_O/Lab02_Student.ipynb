## 📖 Part A: How Do We Measure an Algorithm?

### Why Not Just Use a Stopwatch?

We could time a program in seconds, but the result depends on many things that have nothing to do with the algorithm:

- The speed of the computer (a new laptop vs an old lab PC).
- The programming language (C++ is usually faster than Python).
- Other programs running at the same time.
- The size and type of input.

So instead, we count **how many basic steps** the algorithm performs as a function of the **input size n**. This is machine-independent. A basic step is a simple operation: an assignment, a comparison, an arithmetic operation or accessing a list element.

### Counting Steps: Worked Example

````python
def total_marks(marks):        # n = len(marks)
    total = 0                  # 1 step
    for m in marks:            # loop runs n times
        total = total + m      # 1 step each time -> n steps
    return total               # 1 step
# Total steps ≈ n + 2
print(total_marks([10, 20, 30]))
````

The exact count is **n + 2**. When n is very large (say 1 crore), the “+2” does not matter at all. What matters is that the work **grows in direct proportion to n**. We say this algorithm is **O(n)**, read as “order n” or “Big-O of n”.

### Rules for Finding Big-O

| Rule | Meaning | Example |
|---|---|---|
| 1. Drop constants | Constant factors do not change the growth | 3n + 5 → O(n) |
| 2. Keep the biggest term | For large n, the largest term dominates | n² + 100n + 7 → O(n²) |
| 3. Sequential parts add | Two loops one after another | O(n) + O(n) = O(n) |
| 4. Nested parts multiply | A loop inside a loop | O(n) × O(n) = O(n²) |
| 5. Different inputs, different letters | Two lists of sizes a and b | Nested over both → O(a × b) |

## 📈 Part B: Common Complexity Classes

| Big-O | Name | Example | n = 1,000 → steps |
|---|---|---|---|
| O(1) | Constant | Access `marks[5]`, dictionary lookup | 1 |
| O(log n) | Logarithmic | Binary search (Lab 04) | ≈ 10 |
| O(n) | Linear | Find max, linear search | 1,000 |
| O(n log n) | Linearithmic | Merge sort, quick sort (Lab 07) | ≈ 10,000 |
| O(n²) | Quadratic | Bubble sort, checking all pairs | 1,000,000 |
| O(2ⁿ) | Exponential | Naive recursive Fibonacci (Lab 06) | ≈ 10³⁰⁰ (never finishes) |

> **🧠 Remember the order**
>
> **O(1) < O(log n) < O(n) < O(n log n) < O(n²) < O(2ⁿ) < O(n!)**
> Everything up to O(n log n) is considered efficient for large data. O(n²) is fine only for small inputs. Exponential is practical only for tiny inputs.

### O(1): Constant Time

````python
def get_first(items):
    return items[0]            # one step, whether the list has 5 or 5 crore items

prices = {"sugar": 160, "rice": 320}
print(get_first([4, 8, 15]), prices["rice"])   # dictionary lookup is O(1) on average
````

### O(n): Linear Time

````python
def contains(items, target):
    for x in items:            # in the worst case we check all n items
        if x == target:
            return True
    return False

print(contains([5, 3, 9, 1], 9))
````

### O(n²): Quadratic Time

````python
def has_duplicate_slow(items):
    n = len(items)
    for i in range(n):              # n times
        for j in range(i + 1, n):   # up to n times
            if items[i] == items[j]:
                return True
    return False

print(has_duplicate_slow([3, 1, 4, 1, 5]))
````

The inner loop runs (n−1) + (n−2) + … + 1 = **n(n−1)/2** times. After dropping constants and smaller terms, that is **O(n²)**.

### O(log n): Logarithmic Time

An algorithm is O(log n) when it **cuts the problem in half** at every step. Starting with n = 1,000,000, halving gives 500,000 → 250,000 → … → 1 in only about **20 steps**, because 2²⁰ ≈ 1,000,000.

````python
def halving_steps(n):
    steps = 0
    while n > 1:
        n = n // 2             # problem size is cut in half
        steps += 1
    return steps

for n in [8, 1_000, 1_000_000, 1_000_000_000]:
    print(f"n = {n:>13,} -> {halving_steps(n)} steps")
````

## ⚖️ Part C: Best, Average and Worst Case

The same algorithm can take a different number of steps on different inputs of the same size. Consider searching for a roll number in a list of n students:

| Case | When it happens | Steps | Big-O |
|---|---|---|---|
| Best case | The target is the first element | 1 | O(1) |
| Average case | The target is somewhere in the middle | ≈ n/2 | O(n) |
| Worst case | The target is last or not present | n | O(n) |

In practice we usually report the **worst case**, because it gives a **guarantee**: the algorithm will never be slower than this. When an engineer says “linear search is O(n)”, they mean the worst case.

## 💾 Part D: Space Complexity

**Space complexity** is the extra memory an algorithm needs as the input grows. We usually count **auxiliary space**: memory used in addition to the input itself.

````python
def reverse_new(items):          # O(n) extra space: builds a new list
    result = []
    for x in items:
        result.insert(0, x)
    return result

def reverse_in_place(items):     # O(1) extra space: only two index variables
    left, right = 0, len(items) - 1
    while left < right:
        items[left], items[right] = items[right], items[left]
        left += 1
        right -= 1
    return items

print(reverse_new([1, 2, 3, 4]), reverse_in_place([1, 2, 3, 4]))
````

> **⚖️ The time–space trade-off**
>
> Often you can make a program **faster by using more memory**. Two Sum with a dictionary uses O(n) extra memory to get O(n) time instead of O(n²). Engineers choose based on which resource is more precious.

## ⏱️ Part E: Measuring Time in Python

Theory predicts growth; experiments confirm it. Python’s `time.perf_counter()` gives a precise clock.

````python
import time

def measure(func, data):
    start = time.perf_counter()
    func(data)
    return time.perf_counter() - start

def linear_work(data):
    total = 0
    for x in data:
        total += x
    return total

def quadratic_work(data):
    count = 0
    for x in data:
        for y in data:
            count += 1
    return count

print(f"{'n':>6} {'O(n) seconds':>14} {'O(n^2) seconds':>16}")
for n in [250, 500, 1000, 2000]:
    data = list(range(n))
    print(f"{n:>6} {measure(linear_work, data):>14.6f} {measure(quadratic_work, data):>16.6f}")
````

When n **doubles**, the O(n) time roughly doubles, but the O(n²) time roughly becomes **four times** larger. That is the signature of quadratic growth.

## ⚙️ C++ Corner

Complexity is language-independent: the same loops have the same Big-O in C++. C++ just runs each step faster.

**C++**

```cpp
#include <iostream>
#include <vector>
#include <chrono>
using namespace std;
using namespace std::chrono;

long long quadraticWork(const vector<int>& data) {
    long long count = 0;
    for (int x : data)
        for (int y : data)
            count++;                     // O(n^2)
    return count;
}

int main() {
    for (int n : {1000, 2000, 4000}) {
        vector<int> data(n, 1);
        auto start = high_resolution_clock::now();
        quadraticWork(data);
        auto ms = duration_cast<microseconds>(high_resolution_clock::now() - start).count();
        cout << "n = " << n << " -> " << ms << " microseconds" << endl;
    }
}
```
