## 📖 Part B: What Are Data Structures and Algorithms?

### Data Structure

A **data structure** is a way of **storing and organising data** in a computer so that we can use it efficiently.

### Algorithm

An **algorithm** is a **finite sequence of clear steps** that solves a problem. It is the logic, not the code. The same algorithm can be written in Python, C++ or Java.

#### Five properties every algorithm must have

| Property | Meaning | Tea example |
|---|---|---|
| Input | Zero or more values given from outside | Water, milk, tea leaves, sugar |
| Output | At least one result is produced | A cup of tea |
| Definiteness | Every step is clear, with only one meaning | “Boil for 2 minutes”, not “boil for a while” |
| Finiteness | It must stop after a limited number of steps | The process ends when tea is poured |
| Correctness | It gives the right output for every valid input | Always produces tea, not coffee |

### Worked Example: Find the Highest Marks

**Problem:** A teacher has the marks of students: 67, 89, 45, 92, 78. Find the highest marks.

**Step 1: Pseudocode (language-independent)**

```text
ALGORITHM FindMax(marks)
  INPUT : a list of numbers called marks
  OUTPUT: the largest number

  1. max_value ← marks[0]
  2. FOR each value in marks DO
  3.     IF value > max_value THEN
  4.         max_value ← value
  5. RETURN max_value
```

#### Step 2: Dry run (trace by hand)

````python
def find_max(marks):
    max_value = marks[0]       # step 1: assume first is largest
    for value in marks:        # step 2: check every value
        if value > max_value:  # step 3: found a bigger one?
            max_value = value  # step 4: remember it
    return max_value           # step 5: answer

print(find_max([67, 89, 45, 92, 78]))   # Output: 92
````

The notebook also prints the dry run automatically, so you can compare it with your hand trace:

````python
def find_max_trace(marks):
    max_value = marks[0]
    print(f"Start: max_value = {max_value}")
    for step, value in enumerate(marks, start=1):
        bigger = value > max_value
        if bigger:
            max_value = value
        ans = "Yes" if bigger else "No"
        print(f"Step {step}: value={value:>3} | bigger? {ans:3} | max={max_value}")
    return max_value

find_max_trace([67, 89, 45, 92, 78])
````

> **✅ Check the five properties**
>
> Input: the list of marks. Output: 92. Definite: each line has one meaning. Finite: the loop ends after the last item. Correct: it works for any non-empty list.

````python
def calculate_bill(price, quantity, discount=0):
    """Return total bill after discount (in percent)."""
    total = price * quantity
    return total - (total * discount / 100)

print(calculate_bill(250, 4))        # 1000.0
print(calculate_bill(250, 4, 10))    # 900.0
````

### How Python Stores Data in Memory (Very Important for DSA)

In Python, a variable does **not** contain the value directly. It is a **label (reference)** pointing to an object in memory. Understanding this now will make **linked lists, trees and graphs** much easier later, because those structures are built entirely from references.

````python
a = [1, 2, 3]
b = a              # b points to the SAME list, not a copy
b.append(4)
print(a)           # [1, 2, 3, 4]   a also changed!
print(a is b)      # True: same object in memory

c = a.copy()       # a real copy
c.append(5)
print(a)           # [1, 2, 3, 4]   a is safe
````

| Mutable (can change in place) | Immutable (cannot change) |
|---|---|
| list, dict, set | int, float, str, tuple, bool |

## ⚙️ C++ Corner

For instructors and students working in C++, here is the same Find-Max algorithm. Notice that the **algorithm (logic) is identical**; only the syntax changes.

**C++**

```cpp
#include <iostream>
#include <vector>
using namespace std;

int findMax(const vector<int>& marks) {
    int maxValue = marks[0];
    for (int value : marks) {
        if (value > maxValue)
            maxValue = value;
    }
    return maxValue;
}

int main() {
    vector<int> marks = {67, 89, 45, 92, 78};
    cout << "Highest marks: " << findMax(marks) << endl;   // 92
    return 0;
}
```

## 🔬 Bonus Experiment: Why Data Structures Matter

````python
# Bonus: run after completing the tasks
import random, time

nums = random.sample(range(1, 10_000_000), 5_000)
target = -1                          # impossible target → worst case for both

t0 = time.perf_counter(); two_sum_brute(nums, target); t1 = time.perf_counter()
two_sum_dict(nums, target);          t2 = time.perf_counter()

print(f"Nested loops : {t1 - t0:.4f} s  (~{len(nums) * (len(nums) - 1) // 2:,} comparisons)")
print(f"Dictionary   : {t2 - t1:.4f} s  (~{len(nums):,} lookups)")
print(f"Dictionary is about {(t1 - t0) / max(t2 - t1, 1e-9):.0f}x faster 🚀")
````
